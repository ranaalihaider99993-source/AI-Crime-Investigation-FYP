## Optional approved GPU run with persistent Drive output

Only after CPU verification succeeds, select an appropriate GPU runtime manually
or through authorized automation. Mount Drive through normal Google consent.
The code below never downloads a dataset and creates a new run folder. Existing
production flags, video checkpoints and human reviews remain unchanged.


```python
if not RUN_REVIEWED_IMAGE_EXPERIMENT:
    print('Training disabled. Reviewed image data and explicit execution approval required.')
else:
    from google.colab import drive
    drive.mount('/content/drive')
    import torch
    if not torch.cuda.is_available():
        raise RuntimeError('GPU unavailable; no training started')
    from datetime import datetime, timezone
    from uuid import uuid4
    from scripts.image_evidence_training import train
    project = Path('/content/drive/MyDrive/AI-Crime-Investigation-FYP')
    if not project.is_dir():
        raise RuntimeError('Existing Drive project folder is missing')
    output = project / 'experimental_image_evidence' / (datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ') + '-' + uuid4().hex[:8])
    result = train(IMAGE_MANIFEST, IMAGE_ROOT, output, epochs=5, device='cuda')
    print('Experimental run saved:', output)
    print(result)
```
